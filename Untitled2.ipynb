```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None


        
```

```python
class Stack:

    def __init__(self):
        self.top = None
        self.bottom = None
        self.size = 0


    def push(self,val):

        temp = Node(val)

        if self.top is None:
            self.top = temp
            
        else:
            temp.next = self.top
            self.top = temp 
        self.size += 1
        return temp
    def display_items(self):
        temp = self.top 

        while temp is not None:
            print(temp.val)
            temp = temp.next


    def pop(self):

        if self.size == 0:
            return 'empty stack'

        temp = self.top

        self.top = self.top.next 
        temp.next = None

        self.size -= 1 


        return temp.val
        
    def peek(self):

        if self.size == 0:
            return 'no items on stack'


        return self.top.val


    def isEmpty(self):

        if self.size == 0:
            return True
        else:
            return False

    def delete_all(self):

        for _ in range(self.size):

            self.pop()
            
        
        
        
```

```python
stack = Stack()
```

```python
stack.push(5)
```

```python
stack.push(11)
```
