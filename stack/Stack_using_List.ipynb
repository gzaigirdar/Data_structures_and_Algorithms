```python
class Stack:

    def __init__(self,):
        self.stack = []
        self.size = len(self.stack)

    def pop(self):
        if self.size == 0:
            return "empty stack"
        else:
            self.size -= 1
            return self.stack.pop()
    def push(self,val):

        self.stack.append(val)
        self.size += 1


    def peek(self):

        if self.size == 0:
            return "stack is empty"
        else:
            return self.stack[-1]

    def is_empty(self):

        if self.size == 0:
            return True
        else:
            return False
    def display(self):
        if self.size == 0:
            return "stack is emptty"
            

        for i in reversed(range(len(self.stack))):

            print(self.stack[i])


        
```

```python
stack = Stack()
```

```python
stack.push(1)
```
