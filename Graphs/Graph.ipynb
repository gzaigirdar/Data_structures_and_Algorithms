### Graph definitions and types


Useful definition of graphs:
______________________________

Node or Vertex = A point in a graph is called a vertex.

Edge = A connection between two vertices is called an edge.

Loop = An edge that connects a vertex to itself is called a loop.

Degree of Vertex = Total number of edges connected to a vertex is the degree of the vertex. (A loop counts as two.)

In-Degree of Vertex = In a directed graph, the number of edges coming into a vertex is called its in-degree.

Out-Degree of Vertex = In a directed graph, the number of edges going out from a vertex is called its out-degree.

Adjacency of Vertex = If there is an edge between two vertices, then they are adjacent to each other.

Path = A sequence of vertices and edges connecting two vertices is called a path.

Cycle = A path that starts and ends at the same vertex without repeating any edge or vertex (except the starting/ending vertex) is called a cycle.  

Leaf Vertex = A vertex with a degree of one is called a leaf vertex.

Isolated Vertex = A vertex with no edges connected to it is called an isolated vertex.

Source Vertex = A vertex with in-degree = 0 and out-degree > 0 is called a source vertex.

Sink Vertex = A vertex with out-degree = 0 and in-degree > 0 is called a sink vertex.

**********************************************************



Types of Graphs:
______________________________

Directed Graph = A graph where edges have directions (arrows).  

Undirected Graph = A graph where edges have no directions.  

Directed Acyclic Graph (DAG) = A directed graph with no cycles.  

Weighted Graph = A graph where edges have numeric weights (such as distance, cost, or time).  

Bipartite Graph = A special type of graph where all the vertices can be divided into two sets, and edges only connect vertices from one set to the other (no edge connects vertices within the same set).  

Planar Graph = A graph that can be drawn on a plane without any edges crossing each other.  

Non-Planar Graph = A graph that cannot be drawn on a plane without edges crossing each other.  



```python
from collections import deque

class Graph:
    def __init__(self):
        self.adjacency_list = {}


    def AddVertex(self,vertex):

        if vertex not in self.adjacency_list:
            self.adjacency_list[vertex] = [] 
            return True
        
        return False

    def AddEdge(self,vertex1,vertex2):

        if vertex1 not in self.adjacency_list.keys() or vertex2 not in self.adjacency_list.keys():
            return " none or  one of the vertex doesn't exist"
        else:
            self.adjacency_list[vertex1].append(vertex2)
            self.adjacency_list[vertex2].append(vertex1)

            return True
    
    def RemoveEdge(self,vertex1,vertex2):
        
        if vertex1 not in self.adjacency_list.keys() or vertex2 not in self.adjacency_list.keys():
            return "none or one of the vertices doesn't exist"
        else:
            self.adjacency_list[vertex1].remove(vertex2)
            self.adjacency_list[vertex2].remove(vertex1)
            return True
    
    def RemoveVertex(self,vertex):
        
        if vertex not in self.adjacency_list.keys():
            return "Vertex not found"
        
        else:
            for ver in self.adjacency_list[vertex]:
                self.adjacency_list[ver].remove(vertex)
                
        del self.adjacency_list[vertex]
                
                
                
    def print_list(self):        
        for key in self.adjacency_list.keys(): 
            print(f'Key:{key} edges:{self.adjacency_list[key]}')
            
            
    def BFS(self,vertex):
        visited = set()
        traverse_path = []
        
        queue = deque([])
        
        queue.append(vertex)
        visited.add(vertex)
        
        curr = None 
        while len(queue) > 0:
            
            curr = queue.popleft()
           
           
            traverse_path.append(curr)
            
            
            
            for neighbor in self.adjacency_list[curr]:
                
                
                if neighbor not in visited:
                    queue.append(neighbor)
                    visited.add(neighbor)
        return traverse_path     
            
    def dfs(self,vertex):
        traverse_path = []
        stack = [vertex]
        seen = set()
        

        while stack:
            current  = stack.pop()
            if current not in seen:
                seen.add(current)
                traverse_path.append(current)
                print(current)
                
            for neighbor in self.adjacency_list[current]:
                if neighbor not in seen:
                    stack.append(neighbor)
               
                
        return traverse_path
        
        
        
    
```
