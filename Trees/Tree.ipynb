# Tree definitons



```python
"""
Height of tree       = length of the path from the root to the deepest leaf.
Depth of a node      = number of edges from the root to that node.
Level of a node      = number of edges on the path from the root to that node.
Degree of a node     = number of children of a given node.
Full Binary Tree     = every node has either zero or two children.
Complete Binary Tree = all levels are completely filled except possibly the last level; 
                       the last level is filled from left to right.
Balanced Tree        = all levels are completely filled except possibly the last level;
                       the last level does not have to be filled left to right.
Perfect Binary Tree  = a tree that is both full and balanced.
Leaf                 = a node with no children.
Subtree              = any tree formed by a node and all its descendants.
"""

```

```python
"""
Max nodes at level l:  2^l

Max nodes in height h: 2^(h+1) - 1

Height from nodes (Perfect Binary Tree): h = log2(n + 1) - 1

Leaves in a Full Binary Tree: L = I + 1  (where I = internal nodes) 
--> internal node is a node that's not a leaf
"""
```

# Tree Traversal 

```python
"""
Inorder:   visit left subtree first → visit root → visit right subtree last
           (left → root → right)

Preorder:  visit root first → visit left subtree → visit right subtree last
           (root → left → right)

Postorder: visit left subtree → visit right subtree → visit root last
           (left → right → root)
"""
```

```python
#Inorder  
def Inorder(root_node):
    current = root_node
    if current is None:
        return
    Inorder(current.left_child)
    print(current.value)
    Inorder(current.right_child)

        
```

```python
'''  sample tree
        root
        /   \
        a    b
       /  \
       c   f
''' 
```

```python
# in order (left → root → right)
 
Inorder(root)
```

```python
#Preorder
def Preorder(root):
    current = root

    if current is None:
        return 

    print(current.value)
    Preorder(current.left_child)
    Preorder(current.right_child)
```

```python
# Preorder root->left->right
Preorder(root)
```

```python
# PostOrder 
def PostOrder(root):
    current = root
    if current is None:
        return
    PostOrder(current.left_child)
    PostOrder(current.right_child)
    print(current.value)
```

# PostOrder left->right->root
PostOrder(root)

### level order traversal


```python
from collections import deque
```

```python
# goes each node level by level
def level_order_traversal(root):
    list_of_nodes = []
    # start with adding the root to deque 
    traversal_queue = deque([root])
    while len(traversal_queue) > 0:
        #  pop first current item
        node = traversal_queue.popleft()
        list_of_nodes.append(node.value)
        # append it's children to the queue 
        if node.left_child:
            traversal_queue.append(node.left_child)
        if node.right_child:
            traversal_queue.append(node.right_child)
    return list_of_nodes

print(level_order_traversal(root))
    
```
