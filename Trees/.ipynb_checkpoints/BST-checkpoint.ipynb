### Binary Search Tree definitions and time complexity 

```python
"""
Binary Search Tree (BST) requirements:
- For any node:
    - All nodes in the left subtree have values less than the node's value
    - All nodes in the right subtree have values greater than the node's value
- When inserting a new node:
    - If the value is smaller than the current node, go to the left subtree
    - If the value is greater than the current node, go to the right subtree
"""
 
```

```python
"""
Binary Tree Operations - Time Complexities

1. Traversals (Inorder, Preorder, Postorder, Level-order)
   - Best / Average / Worst: O(n)
   - Explanation: Visit every node once

2. Search in Binary Search Tree (BST)
   - Average: O(log n)
   - Worst (unbalanced): O(n)

3. Insertion in BST
   - Average: O(log n)
   - Worst (unbalanced): O(n)

4. Deletion in BST
   - Average: O(log n)
   - Worst (unbalanced): O(n)

5. Height of tree
   - Binary Tree: O(n) (need to visit all nodes to calculate height)
   - Perfect / Complete / Balanced: O(log n)
"""
```

### Binary Search Tree Class

```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left_child = None
        self.right_child = None
```

```python
from collections import deque
```

```python
class BST:
    def __init__(self):
        self.root = None

    def insert(self,val):
        new_node = TreeNode(val)
        if self.root is None:
            self.root = new_node
            return self.root
        current = self.root

        while current:
            if current.val == new_node.val:
                return None
            if current.val > new_node.val:
                if current.left_child is None:
                    current.left_child = new_node
                    return
                else:
                    current = current.left_child
            else:
                if current.right_child is None:
                    current.right_child = new_node
                    return
                else:
                    current = current.right_child 
                    
    # takes a node and traverses the left subtree to find the min
    def findMin(self,node):
        current = node

        while current.left_child:
            current = current.left_child

        return current 

    # takes a vale and deletes the node with that value 
    def delete_node(self,root, value):
        current = root

        if root is None:
            return None

        # if current val is bigger than the value then move towards left subtree
        if current.val > value:
            current.left_child = self.delete_node(current.left_child,value)

        elif current.val < value:
            # move the right subtree if value bigger than current
            current.right_child = self.delete_node(current.right_child,value)

        else:
            # case one only right child
            # return the right child to function call of parent node of the one we are trying to delete, so right_child get's assigned as child of node's parent
            if current.left_child is None:
                return current.right_child

            # case two only has left child, same procces as before but returns right_child to previouse funtion call
            elif current.right_child is None:
                return current.left_child 

            else:
                ''' 
                case 3: both children exist 
                in the case find the min from right subtree of the node and replace the node to be deleted with that node

                '''
                successor = self.findMin(current.right_child)
                # assign the root value to succssor 
                current.val = successor.val
                # now delete the successor using right subtree of the node to be deleted
                current.right_child = self.delete_node(current.right_child,successor.val)

            return current
    


    # this delete item deletes from the root of the tree
    def delete_item(self, value):
        if self.root is None:
            return 'Tree is empty'
        self.root = self.delete_node(self.root, value)

            
    # max of a tree given a node
    def find_max(self,node):
        current = node

        while current.right_child:
            current = current.right_child

        return current 

    def search(self,val):
        current = self.root

        while current:

            if current.val == val:

                return current 

            elif current.val > val:
                current = current.left_child
            else:
                current = current.right_child
        return None

    def deleteTree(self):

        self.root = None
        return ' the tree has been delete'

    # method uses search functio to see if the value exists and returna boolean depends if the value exist or not
    def contains(self,val):
        return self.search(val) is not None

    def levelOrder_traversal(self):
        list_of_nodes_visited = deque([self.root])
        values_of_nodes = []

        while len(list_of_nodes_visited) > 0:

            current = list_of_nodes_visited.popleft()
            print(current.val)
            print('-->')
            if current.left_child:
                list_of_nodes_visited.append(current.left_child)
            if current.right_child:
                list_of_nodes_visited.append(current.right_child)
    
    def print_tree(self, node=None, level=0):
            if node is None:
                node = self.root
            if node is None:
                return  # Added base case: exit if tree/subtree is empty
            if node.right_child:
                self.print_tree(node.right_child, level + 1)
            print('    ' * level + str(node.val))
            if node.left_child:
                self.print_tree(node.left_child, level + 1)

   
    
         
     # manual add for tree based on existe node

    def add(self, parent_val, child_val):
        parent = self.search(parent_val)
    
        if parent is None:
            return 'Node doesn’t exist'
        else:
            if parent.left_child is None and child_val < parent.val:
                parent.left_child = TreeNode(child_val)
                return parent.left_child
            elif parent.right_child is None and child_val > parent.val:
                parent.right_child = TreeNode(child_val)
                return parent.right_child
            else:
                return 'Parent node doesn’t have space to add this child node'
    def debug_print_tree(self, node=None):
        if node is None:
            node = self.root
        if node is None:
            print("Tree is empty")
            return
        queue = deque([node])
        while queue:
            current = queue.popleft()
            left_val = current.left_child.val if current.left_child else None
            right_val = current.right_child.val if current.right_child else None
            print(f"Node: {current.val}, Left: {left_val}, Right: {right_val}")
            if current.left_child:
                queue.append(current.left_child)
            if current.right_child:
                queue.append(current.right_child)


            
            
            
        

        
            
        


```
